```python
#Bai1
def max_of_two(a, b):
    if a > b:
        return a
    else:
        return b

a, b = map(int, input("nhap hai so:").split())
print(f"so lon hon la: {max_of_two(a, b)}")
```

```python
#Bai2
def swap(a, b):
    a = a ^ b
    b = a ^ b
    a = a ^ b
    return a, b

a, b = map(int, input("nhap hai so:").split())
print(f"hai so sau khi hoan doi la: {swap(a, b)}")
```

```python
#Bai3
def songuyento(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
        else:
            return True
        
n = int(input("nhap so nguyen n:"))
print(f"{n} la so nguyen to: {songuyento(n)}")        
```

```python
#Bai4
def sohoanhao(n):
    sum = 0
    for i in range(1, n):
        if n % i == 0:
            sum += i
    if sum == n:
        return True
    else:
        return False
     
n = int(input("nhap so nguyen n:"))
print(f"{n} la so hoan hao: {sohoanhao(n)}")
```

```python
#Bai5
n = list(map(int, input("nhap danh sach so nguyen:").split()))
m = int(input("nhap so nguyen m:"))
def timkiem(n, m):
    for i in n:
        if i == m:
            return True
    
    return False
print(f"{m} co trong danh sach: {timkiem(n, m)}")
```

```python
#Bai6
def giaithua(n):
    sum = 1
    if n == 0 or n == 1:
        return 1
    else:
        for i in range(2, n+1):
            sum *= i
    
    return sum
n = int(input("nhap so nguyen n:"))
print(f"giai thua cua {n} la: {giaithua(n)}")
```

```python
#Bai7
def tinhtoan(n, m):
    a = n + m
    b = n - m
    c = n * m
    d = n / m
    a, b, c, d = round(a,2), round(b,2), round(c,2), round(d,2)
    return a, b, c, d

n, m = map(float, input("nhap hai so:").split())
print(f"ket qua cua phep toan +, -, *, / lan luot la: {tinhtoan(n, m)}")
```

```python
#Bai8
def khoangcachhamming(a, b):
    result = a ^ b
    count = 0
    while result > 0:
        count += result & 1
        result >>= 1
    
    return count

a, b = map(int, input("nhap hai so nguyen (a b): ").split())
print(f"Khoang cach Hamming giua {a} va {b} la: {khoangcachhamming(a, b)}")
```

```python
#Bai9
def tongchuso(n):
    sum = 0
    n = abs(n)
    while True:
        sum = sum + n % 10
        n  = n // 10
        if n == 0:
            break
    return sum

n = int(input("nhap so nguyen n:"))
print(f"tong chu so cua so {n} la {tongchuso(n)}")
```

```python
#Bai10
def anhxa(a, b):
    if len(a) != len(b):
        return False
    return sorted(a) == sorted(b)

a = input("Nhap tu thu nhat: ")
b = input("Nhap tu thu hai: ")
print(f"'{a}' va '{b}' la dang cau: {anhxa(a, b)}")
```

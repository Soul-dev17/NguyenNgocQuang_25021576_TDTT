```python
#Bai4
import re
lst = list(re.split(r'[: ]', input()))
dic = {}
key_list = []
value_list = []
for i in range(0,len(lst)):
    if i % 2 == 0:
        key_list.append(lst[i])
    else:
        value_list.append(lst[i])

for i in range(0,len(key_list)):
    if key_list[i] not in dic:
        dic[key_list[i]] = []
    dic[key_list[i]].append(value_list[i])
    
print(dic)
```

```python
#Bai5
numbers = list(map(int, input().split()))
p, n, z = 0, 0, 0
for i in numbers:
    if i > 0:
        p += 1
    elif i < 0:
        n += 1
    else:
        z += 1

ketqua = {
    "positives" : p,
    "negatives" : n,
    "zeros" : z
}
print(ketqua)
```

```python
#Bai9
n1 = int(input())
D1 = {}
for _ in range(n1):
    k, v = input().split()
    v = int(v)
    D1[k] = D1.get(k, 0) + v

n2 = int(input())
D2 = {}
for _ in range(n2):
    k, v = input().split()
    v = int(v)
    D2[k] = D2.get(k, 0) + v

merged = D1.copy()
for k, v in D2.items():
    merged[k] = merged.get(k, 0) + v

result = {}
for k in sorted(merged.keys()):
    result[k] = merged[k]
print(result)
```

```python
#Bai10
nums = list(map(int, input().split()))
k = int(input())

pairs = []
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == k:
            pairs.append((nums[i], nums[j]))

pairs.sort(key=lambda t: t[0])
print(pairs)
```

```python
#Bai11
nums = tuple(map(int, input().split()))

even_list = []
odd_list = []
for x in nums:
    if x % 2 == 0:
        even_list.append(x)
    else:
        odd_list.append(x)

print(tuple(even_list))
print(tuple(odd_list))
```

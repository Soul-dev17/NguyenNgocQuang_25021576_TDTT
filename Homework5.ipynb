```python
#Bai16
n = int(input("Nhap so nguyen duong n: "))
chan = 0
le = 0
while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        chan += 1
    else:
        le += 1
    n = n // 10
print(f"So chu so chan: {chan}")
print(f"So chu so le: {le}")
```

```python
#Bai19
def tim_vi_tri_max(lst):
    if len(lst) == 0:
        return -1
    vi_tri_max = 0
    gia_tri_max = lst[0]
    for i in range(1, len(lst)):
        if lst[i] > gia_tri_max:
            gia_tri_max = lst[i]
            vi_tri_max = i
    return vi_tri_max

lst = list(map(int, input("Nhap cac so (cach nhau boi dau cach): ").split()))
print(f"Vi tri co gia tri lon nhat: {tim_vi_tri_max(lst)}")
```

```python
#Bai20
def tim_vi_tri_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1

lst = list(map(int, input("Nhap cac so (cach nhau boi dau cach): ").split()))
k = int(input("Nhap gia tri k can tim: "))
ket_qua = tim_vi_tri_k(lst, k)
if ket_qua == -1:
    print(f"Khong tim thay {k} trong list")
else:
    print(f"Vi tri dau tien cua {k} la: {ket_qua}")
```

```python
#Bai21
chuoi = input("Nhap cac chuoi (cach nhau boi dau cach): ")
danh_sach = chuoi.split()

tu_dien = {}
for i in range(len(danh_sach)):
    if danh_sach[i] in tu_dien:
        tu_dien[danh_sach[i]] = tu_dien[danh_sach[i]] + (i,)
    else:
        tu_dien[danh_sach[i]] = (i,)

print("Dictionary:")
for key, value in tu_dien.items():
    print(f"{key} -> {value}")
```

```python
#Bai23
def xoa_so_nho_hon_x(mang, x):
    ket_qua = []
    for so in mang:
        if so >= x:
            ket_qua.append(so)
    return ket_qua

mang = list(map(int, input("Nhap cac so (cach nhau boi dau cach): ").split()))
x = int(input("Nhap gia tri x: "))
mang_moi = xoa_so_nho_hon_x(mang, x)
print(f"Mang sau khi xoa cac so < {x}: {mang_moi}")
```

```python
#Bai24:
def day_so_0_ve_dau(mang):
    so_luong_0 = mang.count(0)
    mang_khong_0 = [x for x in mang if x != 0]
    return [0] * so_luong_0 + mang_khong_0

mang = list(map(int, input("Nhap cac so (cach nhau boi dau cach): ").split()))
mang_moi = day_so_0_ve_dau(mang)
print(f"Mang sau khi day so 0 ve dau: {mang_moi}")
```

```python
#Bai26
mang = list(map(int, input("Nhap cac so (cach nhau boi dau cach): ").split()))
tong_cuc_dai = 0
for i in range(1, len(mang) - 1):
    if mang[i] > mang[i-1] and mang[i] > mang[i+1]:
        tong_cuc_dai += mang[i]

print(f"Tong cac phan tu cuc dai dia phuong: {tong_cuc_dai}")
```

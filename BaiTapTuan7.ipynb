```python
#Bai1
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

arr = list(map(int, input().split()))
target = int(input())
result = binary_search(arr, 0, len(arr) - 1, target)
print(result)
```

```python
#Bai2
def count_occurrences(arr, x):
    count = 0
    for num in arr:
        if num == x:
            count += 1
    return count

arr = list(map(int, input().split()))
x = int(input())
result = count_occurrences(arr, x)
print(result)
```

```python
#Bai3
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

arr = list(map(int, input().split()))
sorted_arr = bubble_sort(arr)
print(' '.join(map(str, sorted_arr)))
```

```python
#Bai4
def find_most_frequent(arr):
    count_dict = {}
    for num in arr:
        count_dict[num] = count_dict.get(num, 0) + 1
    max_count = max(count_dict.values())
    result = min([num for num, count in count_dict.items() if count == max_count])
    
    return f"{max_count} xuat hien nhieu nhat, som nhat, {max_count} lan"

arr = list(map(int, input().split()))
print(find_most_frequent(arr))
```

```python
#Bai5
def find_pairs_sum(arr, X):
    result = []
    seen = set()
    
    for num in arr:
        complement = X - num
        if complement in seen:
            pair = tuple(sorted([num, complement]))
            if pair not in result:
                result.append(pair)
        seen.add(num)
    
    return len(result)

arr = list(map(int, input().split()))
X = int(input())
print(find_pairs_sum(arr, X))
```

```python
#Bai6
def flatten_list(nested_list):
    result = []
    for item in nested_list:
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result

nested = eval(input())
flattened = flatten_list(nested)
print(flattened)
```

```python
#Bai7
def longest_increasing_subarray(arr):
    if not arr:
        return 0
    
    max_length = 1
    current_length = 1
    
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            current_length += 1
            max_length = max(max_length, current_length)
        else:
            current_length = 1
    
    return max_length

arr = list(map(int, input().split()))

result = longest_increasing_subarray(arr)
print(result)
```

```python
#Bai8
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

intervals = eval(input())
queries = eval(input())

results = []
for q in queries:
    interval_list = []
    for interval in intervals:
        interval_list.extend(range(interval[0], interval[1] + 1))
    
    interval_list.sort()
    result = binary_search(interval_list, q)
    results.append(result)

print(results)
```

```python
#Bai9
def count_smaller(nums):
    result = []
    for i in range(len(nums)):
        count = 0
        for j in range(len(nums)):
            if nums[j] < nums[i]:
                count += 1
        result.append(count)
    return result

nums = list(map(int, input().split()))
result = count_smaller(nums)
print(result)
```

```python
#Bai10
def max_subarray_sum_mod(nums, m):
    max_sum = 0
    for i in range(len(nums)):
        current_sum = 0
        for j in range(i, len(nums)):
            current_sum += nums[j]
            max_sum = max(max_sum, current_sum % m)
    return max_sum

nums = list(map(int, input().split()))
m = int(input())
result = max_subarray_sum_mod(nums, m)
print(result)
```

```python
#Bai11
def count_uet(text):
    words = text.lower().split()
    count = 0
    for word in words:
        clean_word = ''.join(c for c in word if c.isalpha())
        if clean_word == "uet":
            count += 1
    return count

lines = []
while True:
    try:
        line = input()
        lines.append(line)
    except EOFError:
        break

text = ' '.join(lines)
result = count_uet(text)
print(result)
```

```python
#Bai12
courses = {}

while True:
    try:
        line = input().strip()
        if not line:
            break
        
        parts = line.split(',')
        if len(parts) == 2:
            course = parts[0].strip()
            teacher = parts[1].strip()
            
            if course not in courses:
                courses[course] = []
            courses[course].append(teacher)
    except EOFError:
        break

for course in sorted(courses.keys()):
    teachers = sorted(set(courses[course]))
    print(f"{course}: {', '.join(teachers)}")
```

```python
#Bai13
def harmonic_sum(n, r):
    total = 0
    for i in range(1, n + 1):
        total += 1 / (i ** r)
    return total

n = int(input())
r = int(input())
result = harmonic_sum(n, r)
print(result)
```

```python
#Bai1
nums = list(map(int, input().split()))
seen = []
for x in nums:
    if x not in seen:
        seen.append(x)
print(seen)
```

```python
#Bai3
lst = list(map(int, input().split()))
k = int(input())
if len(lst) == 0:
    rotated = []
else:
    k = k % len(lst)
    rotated = lst[k:] + lst[:k]
print(tuple(rotated))
```

```python
#Bai4
parts = input().split()
d = {}
for item in parts:
    if ':' not in item:
        continue
    k, v = item.split(':', 1)
    if k not in d:
        d[k] = []
    d[k].append(v)
print(d)
```

```python
#Bai5
nums = list(map(int, input().split()))
pos = neg = zero = 0
for x in nums:
    if x > 0:
        pos += 1
    elif x < 0:
        neg += 1
    else:
        zero += 1
print({"positives": pos, "negatives": neg, "zeros": zero})
```

```python
#Bai9
n1 = int(input())
d1 = {}
for _ in range(n1):
    k, v = input().split()
    d1[k] = d1.get(k, 0) + int(v)

n2 = int(input())
d2 = {}
for _ in range(n2):
    k, v = input().split()
    d2[k] = d2.get(k, 0) + int(v)

merged = d1.copy()
for k, v in d2.items():
    merged[k] = merged.get(k, 0) + v

result = {}
for k in sorted(merged.keys()):
    result[k] = merged[k]
print(result)
```

```python
#Bai10
nums = list(map(int, input().split()))
k = int(input())
pairs = []
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == k:
            pairs.append((nums[i], nums[j]))
pairs.sort(key=lambda t: t[0])
print(pairs)
```

```python
#Bai11
nums = tuple(map(int, input().split()))
even_list = []
odd_list = []
for x in nums:
    if x % 2 == 0:
        even_list.append(x)
    else:
        odd_list.append(x)
print(tuple(even_list))
print(tuple(odd_list))
```

```python
#Bai12
nums = list(map(int, input().split()))
counts = {}
for x in nums:
    counts[x] = counts.get(x, 0) + 1
max_freq = max(counts.values())
candidates = []
for k, v in counts.items():
    if v == max_freq:
        candidates.append(k)
print(min(candidates))
```

```python
#Bai13
n = int(input())
d = {}
for _ in range(n):
    k, v = input().split()
    d[k] = int(v)
rev = {}
for k, v in d.items():
    rev[v] = k
print(rev)
```

```python
#Bai15
n = int(input())
d = {}
for _ in range(n):
    k_str, v = input().split()
    d[k_str] = int(v)
threshold = int(input())
filtered = {}
for key, val in d.items():
    if val > threshold:
        filtered[key] = val
print(filtered)
```

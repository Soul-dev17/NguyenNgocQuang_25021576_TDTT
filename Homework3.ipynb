```python
#Bai4
n = int(input("nhap so nguyen n: "))
if n < 0:
    n = -n
elif n == 0:
    print("So chu so cua n la: 1")
else:
    count = 0
    temp = n
    while temp > 0:
        temp //= 10
        count += 1
    
    print(f"So chu so cua n la: {count}")
```

```python
#Bai8
a = int(input("nhap so a:"))
b = int(input("nhap so b:"))
counta = 0
countb = 0
for i in range(1, a+1):
    if a % i == 0:
        counta += 1

for j in range(1, b+1):
    if b % j == 0:
        countb += 1
        
count = counta + countb
print(f"tong so uoc cua a va b la: {count}")
```

```python
#Bai9
a = int(input("nhap so a:"))
for i in range(1, a+1):
    if i**2 == a:
        print(f"{a} la so chinh phuong")
        break
    elif i**2 != a and i == a-1:
        print(f"{a} khong phai la so chinh phuong")
```

```python
#Bai10 chua xong
a = int(input("nhap so nguyen duong a:"))
sum = 0
for i in range(1, a+1):
    if a % i == 0:
        sum += i
        continue
    if  i == a:
        if sum == a:
            print(f"{a} la so hoan hao")
        else:
            print(f"{a} khong phai la so hoan hao")
```

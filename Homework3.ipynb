```python
#Bai8
a = int(input("nhap so a:"))
b = int(input("nhap so b:"))
counta = 0
countb = 0
for i in range(1, a+1):
    if a % i == 0:
        counta += 1

for j in range(1, b+1):
    if b % j == 0:
        countb += 1
        
count = counta + countb
print(f"tong so uoc cua a va b la: {count}")
```

```python
#Bai12
n = int(input("nhap so nguyen duong n:"))
while n < 0:
    print("vui long nhap lai so nguyen duong n!")
    n = int(input("nhap so nguyen duong n:"))

s = 0
k = 0
while s + (k + 1) < n:
    k += 1
    s += k

print(k)
```

```python
#Bai13
A = int(input("nhap so nguyen duong A:"))
while A <= 0:
    print("vui long nhap A > 0!")
    A = int(input("nhap so nguyen duong A:"))

s = 0.0
n = 0
while s <= A:
    n += 1
    s += 1.0 / n

print(n)
```

```python
#Bai14
max_val = None
min_val = None

while True:
    x = int(input("nhap so nguyen (nhap -1 de dung): "))
    if x == -1:
        break
    if max_val is None:
        max_val = x
        min_val = x
    else:
        if x > max_val:
            max_val = x
        if x < min_val:
            min_val = x

if max_val is None:
    print("khong co so nao duoc nhap")
else:
    print(f"so lon nhat: {max_val}")
    print(f"so nho nhat: {min_val}")
```

```python
#Bai16
n = int(input("nhap so nguyen duong n: "))
chan = 0
le = 0
while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        chan += 1
    else:
        le += 1
    n = n // 10
print(f"so chu so chan: {chan}")
print(f"so chu so le: {le}")
```

```python
#Bai18
n = int(input("nhap so nguyen duong n: "))
if n < 1:
    print(f"{n} khong phai la so dang 3^k")
else:
    temp = n
    while temp % 3 == 0:
        temp = temp // 3
    if temp == 1:
        print(f"{n} la so dang 3^k")
    else:
        print(f"{n} khong phai la so dang 3^k")
```

```python
#Bai19
A = int(input("nhap so nguyen duong A: "))
fib1 = 1
fib2 = 1
while fib2 <= A:
    fib_next = fib1 + fib2
    if fib_next > A:
        break
    fib1 = fib2
    fib2 = fib_next
print(f"so fibonacci lon nhat khong vuot qua {A} la: {fib2}")
```

```python
#Bai20
n = int(input("nhap so nguyen duong n: "))
count = 0
while n > 0:
    chu_so = n % 10
    if chu_so in [2, 3, 5, 7]:
        count += 1
    n = n // 10
print(f"so chu so la so nguyen to: {count}")
```

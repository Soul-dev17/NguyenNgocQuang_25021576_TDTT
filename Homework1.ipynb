```python
#Bai8
name = input("Ten chu ho: ")
sothangtruoc = int(input("Chi so thang truoc: "))
sothangnay = int(input("Chi so thang nay: "))
so = sothangnay - sothangtruoc
if so <= 50:
    tien = (so * 1984)*1.08
    print("Ho va ten:", name)
    print("Tien phai tra la:", int(tien))
elif 50 < so <= 100:
    tien = ((50 * 1984) + (so - 50) * 2050)*1.08
    print("Ho va ten:", name)
    print("Tien phai tra la:", int(tien))
elif 100 < so <= 200:
    tien = ((50 * 1984) + (50 * 2050) + (so - 100) * 2380)*1.08
    print("Ho va ten:", name)
    print("Tien phai tra la:", int(tien))
elif 200 < so <= 300:
    tien = ((50 * 1984) + (50 * 2050) + (100 * 2380) + (so - 200) * 2998)*1.08
    print("Ho va ten:", name)
    print("Tien phai tra la:", int(tien))
elif 300 < so <= 400:
    tien = ((50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + (so - 300) * 3350)*1.08
    print("Ho va ten:", name)
    print("Tien phai tra la:", int(tien))
elif so > 400:
    tien = ((50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + (100 * 3350) + (so - 400) * 3460)*1.08
    print("Ho va ten:", name)
    print("Tien phai tra la:", int(tien))
```

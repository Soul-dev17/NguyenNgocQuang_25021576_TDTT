```python
#Bai7
n = int(input("nhap so nguyen duong n:"))
number = []
songuyento = []
am = True
while am:
    if n < 2:
        print("vui long nhap lai so nguyen duong lon hon hoac bang 2")
        n = int(input("nhap so nguyen duong n:"))
    else:
        am = False
        
for i in range(2, n+1):
    if n % i == 0:
        number.append(i)

for i in number:
    for j in range(2, i+1):
        if i % j == 0 and j != 1 and j != i:
            break
        else:
            songuyento.append(i)
            break

print(max(songuyento))
```

```python
#Bai8
n = int(input("nhap so nguyen duong n: "))
steps = 0
while True:
    
    temp = n
    reversed_n = 0
    while temp > 0:
        reversed_n = reversed_n * 10 + temp % 10
        temp //= 10
    
    if n == reversed_n:
        break
    
    n = n + reversed_n
    steps += 1

print(f"so buoc da thuc hien: {steps}")
print(f"gia tri palindrome: {n}")
```

```python
#Bai9
n = int(input("nhap so nguyen duong n: "))
result = []
i = 0
while i * i <= n:
    square = i * i
    temp = square
    digits = []
    
    if temp == 0:
        digits.append(0)
    else:
        while temp > 0:
            digit = temp % 10
            digits.append(digit)
            temp //= 10
    
    is_distinct = True
    for j in range(len(digits)):
        for k in range(j + 1, len(digits)):
            if digits[j] == digits[k]:
                is_distinct = False
                break
        if not is_distinct:
            break
    
    if is_distinct:
        result.append(square)
    
    i += 1

if len(result) == 0:
    print("no number")
else:
    for idx in range(len(result)):
        if idx < len(result) - 1:
            print(result[idx], end=" ")
        else:
            print(result[idx])
```

```python
#Bai10
n = int(input("nhap so nguyen duong n: "))
max_length = 0
result_x = 0
for x in range(1, n + 1):
    current = x
    length = 0
    
    while current != 1:
        if current % 2 == 0:
            current = current // 2
        else:
            current = current * 3 + 1
        length += 1
    
    if length > max_length:
        max_length = length
        result_x = x

print(f"{result_x} {max_length}")
```

```python
#Bai13
a = int(input("nhap so nguyen duong a: "))
b = int(input("nhap so nguyen duong b: "))
sum_a = 0
for i in range(1, a):
    if a % i == 0:
        sum_a += i

sum_b = 0
for i in range(1, b):
    if b % i == 0:
        sum_b += i

if sum_a == b and sum_b == a:
    print("true")
else:
    print("false")
```

```python
#Bai7
n = int(input("nhap so nguyen duong n:"))
number = []
songuyento = []
am = True
while am:
    if n < 2:
        print("vui long nhap lai so nguyen duong lon hon hoac bang 2")
        n = int(input("nhap so nguyen duong n:"))
    else:
        am = False
        
for i in range(2, n+1):
    if n % i == 0:
        number.append(i)

for i in number:
    for j in range(2, i+1):
        if i % j == 0 and j != 1 and j != i:
            break
        else:
            songuyento.append(i)
            break

print(max(songuyento))
```

```python
#Bai15
tongcon = int(input("nhap tong so con ga va cho:"))
tongchan = int(input("nhap tong so chan ga va cho:"))
am = True
while am:
    for i in range(1, tongcon):
        ga = i
        cho = tongcon - ga
        if (ga * 2 + cho * 4) == tongchan:
            print(f"so ga la {ga}, so cho la {cho}")
            am = False
```

```python
#Bai7
n = int(input("nhap so nguyen duong n:"))
number = []
songuyento = []
am = True
while am:
    if n < 2:
        print("vui long nhap lai so nguyen duong lon hon hoac bang 2")
        n = int(input("nhap so nguyen duong n:"))
    else:
        am = False
        
for i in range(2, n+1):
    if n % i == 0:
        number.append(i)

for i in number:
    for j in range(2, i+1):
        if i % j == 0 and j != 1 and j != i:
            break
        else:
            songuyento.append(i)
            break

print(max(songuyento))
```

```python
#Bai11
n = int(input("nhap so nguyen n:"))
count = 0
uoc = []
for i in range(1, n+1):
    if n % i == 0:
        uoc.append(i)

for i in uoc:
    if i % 2 == 0:
        count += 1

print(f"so uoc chan cua {n} la: {count}")

```

```python
#Bai1
n = int(input("nhap so nguyen n:"))
sum = 0
for i in range(1, n+1):
    sum += i
print(sum)
```

```python
#Bai2
n = int(input("nhap so nguyen n:"))
am = True
while am:
    if n < 0:
        print("vui long nhap lai so nguyen n!")
        n = int(input("nhap so nguyen n:"))
    else:
        am = False

for i in range(1, n+1):
    if n % i == 0 and i != 1 and i != n:
        print(f"{n} khong la so nguyen to")
        break
    else:
        print(f"{n} la so nguyen to")
        break

```

```python
#Bai6
c = input("nhap mot chu cai thuong (a-z): ")
if len(c) != 1 or not c.isalpha() or not c.islower():
    print("hay nhap mot chu cai thuong")
else:
    unicode_code = ord(c)
    print("ma unicode cua ", c, "la:", unicode_code)
    
    uppercase_char = c.upper()
    print("Ki tu hoa tuong ung la:", uppercase_char)
```

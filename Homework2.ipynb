```python
#Bai3
import math
a = int(input("nhap so a: "))
b = int(input("nhap so b: "))
c = int(input("nhap so c: "))
p = (a + b + c) / 2
dientich = math.sqrt(p * (p - a) * (p - b) * (p - c))
dientich = round(dientich, 2)
if a+b > c and a+c > b and b+c > a:
    print("3 so tren tao thanh canh cua tam giac")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
    print("3 so tren tao thanh canh cua tam giac vuong")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
elif a == b or a == c or b == c:
    print("3 so tren tao thanh canh cua tam giac can")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
elif a == b == c:
    print("3 so tren tao thanh canh cua tam giac deu")
    print("chu vi tam giac la: ", a+b+c)
    print("dien tich tam giac la: ", dientich)
else:
    print("3 so tren khong tao thanh canh cua tam giac")
```

```python
#Bai5
import math
a = int(input("nhap so a: "))
b = int(input("nhap so b: "))
c = int(input("nhap so c: "))
delta = b**2 - 4*a*c
if delta < 0:
    print("phuong trinh vo nghiem")
elif delta == 0:
    x = -b / (2 * a)
    print("phuong trinh co nghiem kep: ", x)
else:
    x1 = (-b + math.sqrt(delta)) / (2 * a)
    x2 = (-b - math.sqrt(delta)) / (2 * a)
    print("phuong trinh co 2 nghiem phan biet: ", x1, "va", x2)
```

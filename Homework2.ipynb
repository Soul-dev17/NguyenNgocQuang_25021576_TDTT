```python
#Bai1
a = int(input("nhap chieu dai hinh chu nhat: "))
b = int(input("nhap chieu rong hinh chu nhat: "))
dientich = a * b
chuvi = (a + b) * 2
print("dien tich hinh chu nhat la: ", dientich)
print("chu vi hinh chu nhat la: ", chuvi)
```

```python
#Bai2
r = int(input("nhap ban kinh hinh tron:"))
chuvi = 2 * r * 3.14
dientich = (r**2) * 3.14
print("chu vi hinh tron la: ", chuvi)
print("dien tich hinh tron la: ", dientich)
```

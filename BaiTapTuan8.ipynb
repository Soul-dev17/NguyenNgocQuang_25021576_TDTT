```python
#Bai16: Kiểm tra mảng đẹp - mảng đẹp là mảng mà mỗi phần tử x xuất hiện đúng x lần
# Nếu không đẹp, xóa các phần tử nhỏ nhất để thành mảng đẹp
arr = list(map(int, input().split()))

# Đếm số lần xuất hiện của mỗi phần tử
count = {}
for num in arr:
    count[num] = count.get(num, 0) + 1

# Kiểm tra mảng có đẹp không
is_nice = True
for num, freq in count.items():
    if num != freq:
        is_nice = False
        break

if is_nice:
    print("YES")
else:
    # Đếm số phần tử cần xóa
    # Nếu số y xuất hiện nhưng không đúng y lần, xóa min(y, số lần xuất hiện)
    remove_count = 0
    for num, freq in count.items():
        if num == 0:
            # Số 0 luôn cần xóa hết vì 0 không thể xuất hiện 0 lần nếu nó có mặt
            remove_count += freq
        elif num != freq:
            # Xóa số phần tử nhỏ hơn: nếu có 5 số 3 thì xóa 3, nếu có 2 số 5 thì xóa 2
            remove_count += min(num, freq)
    
    print(f"NO {remove_count}")
```

```python
#Bai17: Tìm phần tử xuất hiện nhiều hơn n/2 lần - Boyer-Moore Majority Vote
arr = list(map(int, input().split()))
n = len(arr)

# Thuật toán Boyer-Moore
candidate = None
count = 0

for num in arr:
    if count == 0:
        candidate = num
        count = 1
    elif num == candidate:
        count += 1
    else:
        count -= 1

# Kiểm tra lại xem candidate có thực sự xuất hiện > n/2 lần không
if arr.count(candidate) > n / 2:
    print(candidate)
# Nếu không có phần tử nào thỏa mãn, không in gì (hoặc để trống)
```

```python
#Bai18: Sắp xếp - số chẵn trước, số lẻ sau, chẵn tăng dần, lẻ giảm dần
arr = list(map(int, input().split()))

# Tách số chẵn và số lẻ
even = [x for x in arr if x % 2 == 0]
odd = [x for x in arr if x % 2 != 0]

# Sắp xếp chẵn tăng dần, lẻ giảm dần
even.sort()
odd.sort(reverse=True)

# Ghép lại và in kết quả
result = even + odd
print(' '.join(map(str, result)))
```

```python
#Bai22
n = int(input())
names = []
for _ in range(n):
    full_name = input().strip()
    name = full_name.split()[-1]
    names.append(name)

from collections import Counter
count = Counter(names)
max_count = max(count.values())
most_common = [name for name, cnt in count.items() if cnt == max_count]
if len(most_common) > 1:
    most_common.sort(key=lambda x: len(x))

print(most_common[0])
```

```python
#Bai25
s = input().strip()
n = len(s)
k = int(input())
to_remove = n - k
result = []
for digit in s:
    while result and result[-1] < digit and to_remove > 0:
        result.pop()
        to_remove -= 1
    result.append(digit)

result = result[:k]
print(''.join(result))
```

```python
#Bai27
s = input().strip()
stack = []
pairs = {'(': ')', '[': ']', '{': '}'}

for char in s:
    if char in pairs:
        stack.append(char)
    else:
        if not stack or pairs[stack[-1]] != char:
            print("NO")
            break
        stack.pop()
else:
    if not stack:
        print("YES")
    else:
        print("NO")
```

```python
#Bai30
arr = list(map(int, input().split()))
n = len(arr)
dp = [1] * n
parent = [-1] * n
for i in range(1, n):
    for j in range(i):
        if arr[j] < arr[i] and dp[j] + 1 > dp[i]:
            dp[i] = dp[j] + 1
            parent[i] = j

max_length = max(dp)
max_index = dp.index(max_length)
result = []
idx = max_index
while idx != -1:
    result.append(arr[idx])
    idx = parent[idx]
result.reverse()

print(' '.join(map(str, result)))
```
